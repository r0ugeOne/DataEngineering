# Pandas Basics

A comprehensive introduction to pandas DataFrames, data cleaning, and analysis.

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

# Display settings
pd.set_option('display.max_columns', None)
pd.set_option('display.max_rows', 100)
```

## Creating DataFrames

```python
# Create a DataFrame from a dictionary
data = {
    'id': [1, 2, 3, 4, 5],
    'name': ['Alice', 'Bob', 'Charlie', 'Diana', 'Eve'],
    'age': [25, 30, 35, 28, 32],
    'salary': [50000, 60000, 75000, 55000, 65000]
}

df = pd.DataFrame(data)
print(df)
```

```python
# DataFrame info
print(df.info())
print("\nBasic statistics:")
print(df.describe())
```

## Data Selection and Filtering

```python
# Select single column
print(df['name'])

# Select multiple columns
print("\nMultiple columns:")
print(df[['name', 'age']])
```

```python
# Filter rows
print("Ages > 28:")
print(df[df['age'] > 28])

# Multiple conditions
print("\nAge > 28 AND salary > 60000:")
print(df[(df['age'] > 28) & (df['salary'] > 60000)])
```

## Data Cleaning

```python
# Create data with missing values
dirty_data = {
    'id': [1, 2, 3, 4, 5],
    'value': [10, np.nan, 30, 40, np.nan],
    'category': ['A', 'B', None, 'A', 'B']
}

df_dirty = pd.DataFrame(dirty_data)
print("Original:")
print(df_dirty)

# Drop NaN
print("\nAfter dropna():")
print(df_dirty.dropna())
```

```python
# Fill NaN with specific values
print("Fill with 0:")
print(df_dirty.fillna(0))

# Forward fill
print("\nForward fill:")
print(df_dirty.fillna(method='ffill'))
```

## GroupBy and Aggregations

```python
# Create sample data with categories
transactions = pd.DataFrame({
    'user_id': [1, 1, 2, 2, 3, 3, 3],
    'amount': [100, 50, 200, 150, 75, 125, 100],
    'category': ['A', 'B', 'A', 'A', 'B', 'B', 'A']
})

print("Transactions:")
print(transactions)
```

```python
# GroupBy operations
print("Sum by user:")
print(transactions.groupby('user_id')['amount'].sum())

print("\nMultiple aggregations:")
print(transactions.groupby('user_id').agg({
    'amount': ['sum', 'mean', 'count']
}))
```

## Merging and Joining

```python
# Create two DataFrames to merge
users = pd.DataFrame({
    'user_id': [1, 2, 3],
    'name': ['Alice', 'Bob', 'Charlie']
})

orders = pd.DataFrame({
    'user_id': [1, 2, 2, 3],
    'order_amount': [100, 50, 75, 200]
})

print("Users:")
print(users)
print("\nOrders:")
print(orders)
```

```python
# Inner join
print("Inner join:")
print(pd.merge(users, orders, on='user_id'))

# Left join
print("\nLeft join:")
print(pd.merge(users, orders, on='user_id', how='left'))
```
